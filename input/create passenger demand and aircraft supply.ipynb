```python
import numpy as np


def sec_to_ms(sec):
    if type(sec) not in [int, float, np.float64, np.float32, np.int_, np.array]:
        raise TypeError("Can't convert sec to ms. Duration (sec) format is not correct")
    else:
        return sec * 1000


def min_to_sec(min):
    if type(min) not in [int, float, np.float64, np.float32, np.int_]:
        raise TypeError("Can't convert min to sec. Duration (min) format is not correct")
    else:
        return min * 60
```

```python
import scipy.stats as stats
import numpy as np
from typing import Dict, Union, List
import random


def pick_number_from_distribution(population, probabilities):
    """
    Picks a number from a distribution given population and probabilities.
    :param population: np.ndarray of values in the distribution
    :param probabilities: np.ndarray of probabilities of values in the distribution
    :return:
    """
    time = min_to_sec(random.choices(population,
                                     probabilities)[0])
    time_sec = round(max(0, time))
    return sec_to_ms(time_sec)


class DistributionGenerator:
    def __init__(self, distribution_params: Dict, max_val_in_dist: Union[float, int]):
        self.distribution_params = distribution_params
        self.max_val_in_dist = max_val_in_dist
        self.population, self.probabilities = self.generate_distribution()

    def generate_distribution(self) -> [np.ndarray, np.ndarray]:
        """
        Generates a distribution given distribution name and parameters.
        :param distribution_params
        :param max_val_in_dist
        :return:
        population: np.ndarray of values in the distribution
        probabilities: np.ndarray of probabilities of values in the distribution
        """
        dist = getattr(stats, self.distribution_params['distribution_name'])
        population = np.linspace(0, self.max_val_in_dist, 1000)
        probabilities = dist.pdf(x=population, **self.distribution_params['parameters'])
        return population, probabilities

    def pick_number_from_distribution(self):
        """
        Picks a number from the distribution.
        :return:
        """
        return pick_number_from_distribution(self.population, self.probabilities)
```

```python
def cumulative_sum_array(array: Union[np.ndarray, List]) -> np.ndarray:
    """
    Cumulative sum of an array.
    :param array:
    :return: np.ndarray
    """
    return np.cumsum(array)
```

```python
import pandas as pd
import numpy as np

def haversine_vectorize(lon1, lat1, lon2, lat2):
    
    #lon1, lat1, lon2, lat2 = map(np.radians, [lon1, lat1, lon2, lat2])

    newlon = lon2 - lon1
    newlat = lat2 - lat1

    haver_formula = np.sin(newlat/2.0)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(newlon/2.0)**2

    dist = 2 * np.arcsin(np.sqrt(haver_formula ))
    km = 6367 * dist #6367 for distance in KM for miles use 3958
    return km


def calc_vert_to_vert_dist(trips, x_vert_o, y_vert_o,
                           x_vert_d, y_vert_d): 
    """This function calculates vert_to_vert distance and adds a column to the 
    given data file. It then saves the file at the fileLocation.

    Parameters
    ----------
    trips : Pandas dataFrame
        Demand file with origin and destination vertiport locations
    x_vert_o : str
        Column name of the longitude of origin vertiport
    y_vert_o : str
        Column name of the latitude of origin vertiport
    x_vert_d : str
        Column name of the longitude of destination vertiport
    y_vert_d : str
        Column name of the latitude of destination vertiport        
    
    """

    trips['rad_lon_o_vert'] = np.radians(trips.apply(lambda p: p.x_vert_o, axis=1)) #np.radians(trips_copy_gdf_final.centroid_x. df.point_object.apply(lambda p: p.x)
    trips['rad_lat_o_vert'] = np.radians(trips.apply(lambda p: p.y_vert_o, axis=1)) #np.radians(trips_copy_gdf_final.centroid_x. df.point_object.apply(lambda p: p.x)
    trips['rad_lon_d_vert'] = np.radians(trips.apply(lambda p: p.x_vert_d, axis=1)) #np.radians(trips_copy_gdf_final.centroid_x. df.point_object.apply(lambda p: p.x)
    trips['rad_lat_d_vert'] = np.radians(trips.apply(lambda p: p.y_vert_d, axis=1)) #np.radians(trips_copy_gdf_final.centroid_x. df.point_object.apply(lambda p: p.x)

    trips['vert_to_vert_dist'] = haversine_vectorize(trips['rad_lon_o_vert'],trips['rad_lat_o_vert'],\
                                          trips['rad_lon_d_vert'], trips['rad_lat_d_vert'])
    return trips
```

```python
import numpy as np
from math import sin, cos, sqrt, atan2, radians


def calc_haversine_dist(lat1, lon1, lat2, lon2):
    # 6367 for distance in KM for miles use 3958
    r = 3958

    lat1 = np.radians(lat1)
    lon1 = np.radians(lon1)
    lat2 = np.radians(lat2)
    lon2 = np.radians(lon2)

    dlon = lon2 - lon1
    dlat = lat2 - lat1

    a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
    c = 2 * atan2(sqrt(a), sqrt(1 - a))

    return round(r * c)


def calc_destination_vertiport_distances(vertiport_network):
    vertiport_distances = {}
    for _, row in vertiport_network.iterrows():
        vertiport_distances[row['vertiport_id']] = calc_haversine_dist(row['o_vert_lat'], 
                                                                       row['o_vert_lon'], 
                                                                       row['d_vert_lat'], 
                                                                       row['d_vert_lon'])
    return vertiport_distances
```

```python
def map_nested_dict_keys_to_values_from_list(dictionary, list_of_keys, last_key):
    """
    Map nested dictionary keys to values from a list input.
    :param dictionary:
    :param list_of_keys:
    :param last_key:
    :return:
    """
    return [dictionary[key][last_key] for key in list_of_keys]
```

```python
def nested_dict_filter_keys(d, depth, key):
    """
    Usage:
        list(nested_dict_keys({1: {2: 3, 4: 5}}, 2))
        list(nested_dict_keys(vertipad_PARK,2))
    """
    if depth == 1:
        for i in d.keys():
            if i == key:
                yield d[key]
    else:
        for v in d.values():
            if isinstance(v, dict):
                for i in nested_dict_filter_keys(v, depth - 1, key):
                    yield i
```
